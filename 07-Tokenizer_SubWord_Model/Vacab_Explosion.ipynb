# What is Vacab Explosion?
 Vacab explosion is a phenomenon where a fixed size vocabulary of a model is not able to capture the vast number of words in a natural language. This results in out-of-vocabulary (OOV) words which are not present in the vocabulary of the model. This can lead to a significant degradation in the performance of the model. This is because the model is not able to capture the semantics of the words that are not present in its vocabulary.


The working of vocab explosion can be explained as follows:

* The model is initially trained on a small dataset which results in a small vocabulary.
* As the model is applied to a larger dataset, it encounters a large number of out-of-vocabulary (OOV) words.
* The model is unable to capture the semantics of the OOV words because they are not present in its vocabulary.
* The model is forced to map the OOV words to a special token such as "<unk>".
* This results in a significant degradation in the performance of the model.



##### The problem with vacab explosion is that the model is unable to capture the semantics of the out-of-vocabulary (OOV) words. This results in a significant degradation in the performance of the model.
