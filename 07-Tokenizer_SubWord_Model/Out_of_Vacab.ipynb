# What is Out Of Vacab
Out of Vacab (OOV) is a measure of whether a particular word is present in a given vocabulary or not. It is commonly used in NLP tasks such as text classification, language modeling, and topic modeling.

OOV is usually calculated as the ratio of the number of out-of-vocabulary words to the total number of words in a given text.

OOV = (Number of out-of-vocabulary words) / (Total number of words)


The working of OOV can be explained as follows:

* The model is initially trained on a small dataset which results in a small vocabulary.
* As the model is applied to a larger dataset, it encounters a large number of out-of-vocabulary (OOV) words.
* The model is unable to capture the semantics of the OOV words because they are not present in its vocabulary.
* The model is forced to map the OOV words to a special token such as "<unk>".
* This results in a significant degradation in the performance of the model.
