
## What is Semantic Neighbourhood

Semantic neighbourhood is a type of visualization that is used to show the semantic relationships between words in a text document. It is a powerful tool for understanding the meaning of text data and for identifying patterns and relationships in the data.

Semantic neighbourhood is a type of graph visualization that is used to represent the semantic relationships between words in a text document. It is typically used in natural language processing tasks such as text classification, sentiment analysis, and named entity recognition.

semantic neighbourhood can be used to visualize the relationships between words in a text document, such as:

* Co-occurrence: The frequency with which two words appear together in a text document.
* Collocation: The frequency with which two words appear near each other in a text document.
* Semantic similarity: The similarity in meaning between two words in a text document.

semantic neighbourhood can be calculated using various algorithms such as:

* Word2Vec
* Doc2Vec
* Gensim
* Spacy
