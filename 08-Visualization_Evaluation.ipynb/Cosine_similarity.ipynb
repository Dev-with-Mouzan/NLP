
# What is Cosine Similarity Score
\
Cosine similarity is a measure of the cosine angle between two non-zero vectors. It is a measure of the similarity between two vectors, and is defined as the dot product of the two vectors divided by the product of their magnitudes. Cosine similarity is a measure of the orientation of two vectors, and is widely used in natural language processing and information retrieval. It is often used to compare the similarity of two documents, or to compare the similarity of a document and a query. The cosine similarity score can range from -1 to 1, where a score of 1 means that the two vectors are identical, and a score of -1 means that the two vectors are orthogonal. A score of 0 means that the two vectors are independent of each other.

# Example of Cosine Similarity


```python
def cosine_similarity(doc1, doc2):
    import numpy as np
    from sklearn.metrics.pairwise import cosine_similarity
    doc1 = np.array(doc1)
    doc2 = np.array(doc2)
    return cosine_similarity(doc1.reshape(1, -1), doc2.reshape(1, -1))[0][0]

document1 = [1, 0, 0, 1, 0]
document2 = [0, 0, 0, 1, 0]
similarity = cosine_similarity(document1, document2)
print(f"Cosine Similarity: {similarity}")

# Output: Cosine Similarity: 0.7071067811865475
```
