### Pre-trained Embedding Models

Pre-trained embedding models are models that are trained on large amount of text data and are publicly available. These models are often used as a starting point for training a model on a specific task.

Some popular pre-trained embedding models include:

* Word2Vec
* GloVe
* FastText
* BERT
* RoBERTa

These models are trained on large datasets such as Wikipedia, books, and web pages. They are often released with pre-trained weights which can be used as a starting point for training a model on a specific task.

For example, the pre-trained Word2Vec model is trained on a dataset of 100 billion words from Google News. The pre-trained GloVe model is trained on a dataset of 840 billion tokens from Common Crawl.

Pre-trained embedding models have several advantages:

* They are often more accurate than randomly initialized models
* They require less training data
* They can be fine-tuned for specific tasks

However, pre-trained embedding models also have some disadvantages:

* They require a large amount of computational resources
* They may not perform well on tasks that are very different from the tasks they were trained on
* They may require a large amount of labelled data to fine-tune.
