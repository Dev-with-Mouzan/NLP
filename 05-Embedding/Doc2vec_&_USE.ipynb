```python
d="""1. The weather is cold today.
2. I forgot my jacket at home.
3. The laptop battery died during the meeting.
4. He bought a new gaming laptop yesterday.
5. The restaurant near my house makes great pizza.
6. I tried a new pasta recipe last night.
7. She is studying machine learning for her final exam.
8. Neural networks require a lot of data to train properly.
9. I woke up late and missed the bus.
10. He usually walks to work every morning.
11. The stock market dropped sharply this week.
12. Investors are worried about rising inflation.
13. I watched a science fiction movie last night.
14. The new superhero film broke box office records.
15. The gym was completely packed this evening.
16. I started a new fitness program last month.
17. My phone screen cracked after it fell on the floor.
18. He bought a new phone because his old one stopped working.
19. The cat slept on the sofa the entire afternoon.
20. My dog loves running in the park on weekends.
"""
```

```python
import nltk
nltk.download("punkt")
nltk.download("wordnet")
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
import re
d=re.sub(r"[^a-z/s]"," ",d.lower())
```

```python
d
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
token=word_tokenize(d)
```

```python
stop=stopwords.words("english")
```

```python
token=[w for w in token if w not in stop]
```

```python
token
```

```python
from gensim.models.doc2vec import Doc2Vec, TaggedDocument
```

```python
tg_doc = [TaggedDocument(words=token, tags=["main_document"])]
```

```python
model=Doc2Vec(tg_doc,vector_size=50,min_count=1,window=2,epochs=40)
```

```python
model.build_vocab(tg_doc)
```

```python
model.train(tg_doc, total_examples=model.corpus_count, epochs=model.epochs)
```

```python
print(model.infer_vector(["weather", "cold"]))
```

```python
import re
from nltk.tokenize import word_tokenize

def preprocess(text):
    text = text.lower()
    text = re.sub(r"[^a-z\s]", " ", text)
    return word_tokenize(text)
```

```python
tokenized_docs = [preprocess(s) for s in d]
```

```python
!pip install gensim
```

```python
from gensim.models.doc2vec import Doc2Vec, TaggedDocument

tagged_docs = [
    TaggedDocument(words=tokens, tags=[str(i)])
    for i, tokens in enumerate(tokenized_docs)
]
```

```python
model = Doc2Vec(vector_size=100, window=5, min_count=1, workers=4, epochs=50)

model.build_vocab(tagged_docs)
model.train(tagged_docs, total_examples=model.corpus_count, epochs=model.epochs)
```
