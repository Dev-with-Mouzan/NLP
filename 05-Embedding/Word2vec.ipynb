
# What is Word2vec?

Word2vec is a popular unsupervised machine learning algorithm used for natural language processing. It is an extension of the Word Embedding concept, which is a method of representing words as vectors in a high-dimensional space. Word2vec is used to capture the semantic relationships between words.

Word2vec works by training a model on a large corpus of text data and then using that model to predict the context words for a given target word. The goal is to find the words that are most likely to appear in the same context as the target word.

Word2vec uses two main algorithms to train the model: Continuous Bag of Words (CBOW) and Skip-Gram. CBOW predicts a target word based on its context, while Skip-Gram predicts the context words based on a target word.

Word2vec provides several advantages over traditional methods of natural language processing, such as:

* Capturing semantic relationships between words
* Handling out-of-vocabulary words
* Improving the accuracy of natural language processing tasks

Word2vec is widely used in many natural language processing tasks, such as:

* Text classification
* Sentiment analysis
* Named entity recognition
* Question answering
* Machine translation



## Architecture of Word2vec

The architecture of Word2vec consists of three main components:

* Input Layer: This layer takes in the input words and converts them into one-hot encoded vectors.
* Hidden Layer: This layer takes in the one-hot encoded vectors and passes them through a fully connected neural network to produce the output vectors.
* Output Layer: This layer takes in the output vectors and produces the final word embeddings.

The Word2vec model is trained using the CBOW and Skip-Gram algorithms. The CBOW algorithm predicts the target word based on its context, while the Skip-Gram algorithm predicts the context words based on the target word. The training process involves minimizing the loss function between the predicted and actual output vectors.

The Word2vec model is typically trained on a large corpus of text data and produces high-dimensional word embeddings. The word embeddings capture the semantic relationships between words and are used as input features for natural language processing tasks.

```python
from gensim.models import Word2Vec
```

```python
sentences = [
    ["i", "love", "machine", "learning"],
    ["deep", "learning", "is", "amazing"],
    ["i", "love", "deep", "learning"],
    ["machine", "learning", "drives", "ai"],
    ["ai", "is", "the", "future"]
]
```

```python
model = Word2Vec(sentences, vector_size=4, window=3, min_count=1, sg=0)  
```

```python
print("\nVector for 'learning':\n", model.wv['learning'])
print("\nMost similar to 'learning':\n", model.wv.most_similar('learning'))
```
