
# What are Sentence or Document Embeddings?

Sentence or Document Embeddings are a way of representing sentences or documents as vectors in a high-dimensional space. The goal is to capture the meaning of the text in a way that can be used in machine learning models.

Sentence embeddings are used to represent individual sentences as vectors in a high-dimensional space. This allows for the capture of semantic relationships between sentences.

Document embeddings are used to represent entire documents as vectors in a high-dimensional space. This allows for the capture of semantic relationships between documents.

Sentence and document embeddings are widely used in natural language processing tasks such as:

* Text classification
* Sentiment analysis
* Named entity recognition
* Question answering
* Machine translation


## What is Doc2Vector?

Doc2Vector is a type of document embedding model that is widely used in natural language processing tasks. It is an extension of the Word2Vec model, which is used to represent individual words as vectors in a high-dimensional space.

Doc2Vector uses a combination of the Continuous Bag of Words (CBOW) and Skip-Gram algorithms to learn the document embeddings. The model is trained on a corpus of text data and produces high-dimensional vectors that capture the semantic relationships between documents.

## What is Universal Sentence Encoder?

Universal Sentence Encoder is a type of document embedding model that is widely used in natural language processing tasks. It is based on the Doc2Vector model and uses a combination of the Continuous Bag of Words (CBOW) and Skip-Gram algorithms to learn the document embeddings.

Universal Sentence Encoder is a variant of the Doc2Vector model that is trained on a larger corpus of text data and produces higher-dimensional vectors that capture more semantic relationships between documents.
